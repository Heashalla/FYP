## Notebook 01: Raw Data Extraction

This notebook imports and inspects the original NHANES XPT files for the 2013–2014, 2015–2016 and 2017–2018 survey cycles.

```python
from pathlib import Path

project_folder = Path("/content/drive/MyDrive/NHANES_Project")
zip_file = project_folder / "NHANES_raw_data.zip"

print("Project folder:", project_folder)
print("Project folder exists:", project_folder.exists())
print()
print("ZIP file:", zip_file)
print("ZIP file exists:", zip_file.exists())

if zip_file.exists():
    size_mb = zip_file.stat().st_size / (1024 * 1024)
    print(f"ZIP file size: {size_mb:.2f} MB")
    print("\nStatus: ZIP file located successfully.")
else:
    print("\nStatus: ZIP file was not found.")
```

```python
import zipfile
from pathlib import Path

with zipfile.ZipFile(zip_file, "r") as zip_archive:
    all_entries = zip_archive.namelist()

# Exclude folder entries
files_inside = [
    entry for entry in all_entries
    if not entry.endswith("/")
]

# Select XPT files regardless of upper/lowercase extension
xpt_files = [
    entry for entry in files_inside
    if Path(entry).suffix.lower() == ".xpt"
]

other_files = [
    entry for entry in files_inside
    if Path(entry).suffix.lower() != ".xpt"
]

print("ZIP inspection completed")
print("-" * 60)
print("Total files found:", len(files_inside))
print("XPT files found:", len(xpt_files))
print("Other files found:", len(other_files))

print("\nXPT FILES")
print("-" * 60)

for number, file_name in enumerate(sorted(xpt_files), start=1):
    print(f"{number:02d}. {file_name}")

if other_files:
    print("\nOTHER FILES")
    print("-" * 60)

    for file_name in sorted(other_files):
        print(file_name)

print("\nVALIDATION")
print("-" * 60)

if len(xpt_files) == 30:
    print("Status: All 30 expected XPT files were found.")
else:
    print(f"Status: Expected 30 XPT files, but found {len(xpt_files)}.")
```

```python
expected_files = {
    "2013_2014": {
        "BMX_H.XPT", "BPQ_H.XPT", "BPX_H.XPT", "DEMO_H.XPT",
        "DIQ_H.XPT", "DR1TOT_H.XPT", "GLU_H.XPT", "HDL_H.XPT",
        "PAQ_H.XPT", "TRIGLY_H.XPT"
    },
    "2015_2016": {
        "BMX_I.XPT", "BPQ_I.XPT", "BPX_I.XPT", "DEMO_I.XPT",
        "DIQ_I.XPT", "DR1TOT_I.XPT", "GLU_I.XPT", "HDL_I.XPT",
        "PAQ_I.XPT", "TRIGLY_I.XPT"
    },
    "2017_2018": {
        "BMX_J.XPT", "BPQ_J.XPT", "BPX_J.XPT", "DEMO_J.XPT",
        "DIQ_J.XPT", "DR1TOT_J.XPT", "GLU_J.XPT", "HDL_J.XPT",
        "PAQ_J.XPT", "TRIGLY_J.XPT"
    }
}

all_cycles_valid = True

for cycle, expected in expected_files.items():

    found = {
        Path(file_name).name.upper()
        for file_name in xpt_files
        if cycle in file_name
    }

    missing = expected - found
    unexpected = found - expected

    print(f"\nCycle: {cycle}")
    print("-" * 40)
    print("Expected files:", len(expected))
    print("Files found:", len(found))

    if not missing and not unexpected:
        print("Status: Complete")
    else:
        all_cycles_valid = False
        print("Status: Check required")

        if missing:
            print("Missing files:", sorted(missing))

        if unexpected:
            print("Unexpected files:", sorted(unexpected))

print("\nOVERALL VALIDATION")
print("-" * 40)

if all_cycles_valid:
    print("Status: All three survey cycles are complete.")
    print("The ZIP file is ready for extraction.")
else:
    print("Status: One or more cycles require correction.")
```

```python
import zipfile
from pathlib import Path

extraction_folder = Path("/content/NHANES_Project")
extraction_folder.mkdir(parents=True, exist_ok=True)

with zipfile.ZipFile(zip_file, "r") as zip_archive:
    zip_archive.extractall(extraction_folder)

raw_data_folder = extraction_folder / "raw_data"

extracted_xpt_files = sorted([
    file_path
    for file_path in raw_data_folder.rglob("*")
    if file_path.is_file() and file_path.suffix.lower() == ".xpt"
])

print("Extraction completed")
print("-" * 60)
print("Extraction folder:", extraction_folder)
print("Raw-data folder exists:", raw_data_folder.exists())
print("XPT files extracted:", len(extracted_xpt_files))

print("\nEXTRACTED FILES")
print("-" * 60)

for number, file_path in enumerate(extracted_xpt_files, start=1):
    relative_path = file_path.relative_to(extraction_folder)
    print(f"{number:02d}. {relative_path}")

print("\nVALIDATION")
print("-" * 60)

if len(extracted_xpt_files) == 30:
    print("Status: All 30 XPT files were extracted successfully.")
else:
    print(f"Status: Expected 30 files, but extracted {len(extracted_xpt_files)}.")
```

```python
import pandas as pd
from pathlib import Path
from openpyxl.styles import Font, PatternFill, Alignment

# Permanent Excel output location in Google Drive
excel_root = project_folder / "raw_excel"
excel_root.mkdir(parents=True, exist_ok=True)

extraction_summary = []

print("Starting XPT-to-Excel conversion")
print("=" * 70)

for number, xpt_path in enumerate(extracted_xpt_files, start=1):

    cycle = xpt_path.parent.name

    cycle_output_folder = excel_root / cycle
    cycle_output_folder.mkdir(parents=True, exist_ok=True)

    excel_name = f"{xpt_path.stem.upper()}.xlsx"
    excel_path = cycle_output_folder / excel_name

    try:
        # Read the original XPT file
        dataframe = pd.read_sas(
            xpt_path,
            format="xport",
            encoding="latin-1"
        )

        # Display participant IDs as whole numbers in Excel
        if "SEQN" in dataframe.columns:
            dataframe["SEQN"] = pd.to_numeric(
                dataframe["SEQN"],
                errors="coerce"
            ).astype("Int64")

        duplicate_seqn = (
            int(dataframe["SEQN"].duplicated().sum())
            if "SEQN" in dataframe.columns
            else None
        )

        # Export to Excel
        with pd.ExcelWriter(
            excel_path,
            engine="openpyxl"
        ) as writer:

            dataframe.to_excel(
                writer,
                sheet_name="Raw_Data",
                index=False
            )

            worksheet = writer.book["Raw_Data"]

            # Keep the column names visible when scrolling
            worksheet.freeze_panes = "A2"

            # Add filters to the complete table
            worksheet.auto_filter.ref = worksheet.dimensions

            # Format the header row
            header_fill = PatternFill(
                fill_type="solid",
                fgColor="1F4E78"
            )

            for cell in worksheet[1]:
                cell.font = Font(
                    bold=True,
                    color="FFFFFF"
                )
                cell.fill = header_fill
                cell.alignment = Alignment(
                    horizontal="center"
                )

        extraction_summary.append({
            "Survey_Cycle": cycle,
            "XPT_File": xpt_path.name,
            "Excel_File": excel_name,
            "Rows": dataframe.shape[0],
            "Columns": dataframe.shape[1],
            "Duplicate_SEQN": duplicate_seqn,
            "Missing_Cells": int(dataframe.isna().sum().sum()),
            "Status": "Successfully converted"
        })

        print(
            f"{number:02d}/30 | {cycle} | "
            f"{xpt_path.name} → {excel_name} | "
            f"{dataframe.shape[0]:,} rows × "
            f"{dataframe.shape[1]} columns"
        )

    except Exception as error:

        extraction_summary.append({
            "Survey_Cycle": cycle,
            "XPT_File": xpt_path.name,
            "Excel_File": excel_name,
            "Rows": None,
            "Columns": None,
            "Duplicate_SEQN": None,
            "Missing_Cells": None,
            "Status": f"Failed: {error}"
        })

        print(
            f"{number:02d}/30 | FAILED | "
            f"{xpt_path.name} | {error}"
        )

print("=" * 70)
print("Conversion process completed.")
```

```python
summary_dataframe = pd.DataFrame(extraction_summary)

reports_folder = project_folder / "reports"
reports_folder.mkdir(parents=True, exist_ok=True)

summary_file = reports_folder / "extraction_summary.xlsx"

summary_dataframe.to_excel(
    summary_file,
    sheet_name="Extraction_Summary",
    index=False
)

successful_files = (
    summary_dataframe["Status"] == "Successfully converted"
).sum()

failed_files = len(summary_dataframe) - successful_files

print("EXTRACTION SUMMARY")
print("-" * 70)
print("Files processed:", len(summary_dataframe))
print("Successfully converted:", successful_files)
print("Failed conversions:", failed_files)
print("Excel output folder:", excel_root)
print("Summary report:", summary_file)

display(summary_dataframe)
```

```python
dataset_register = pd.DataFrame([
    ["DEMO", "Demographics and survey design", "Age, sex, survey weights, strata and PSU", "Primary"],
    ["BMX", "Body measurements", "Height, weight, waist circumference and BMI", "Primary"],
    ["BPX", "Blood-pressure examination", "Repeated systolic and diastolic readings", "Primary"],
    ["BPQ", "Blood-pressure questionnaire", "Hypertension history and treatment", "Primary"],
    ["GLU", "Plasma fasting glucose", "Fasting glucose and fasting-subsample weight", "Primary"],
    ["DIQ", "Diabetes questionnaire", "Insulin and diabetic-pill use", "Primary"],
    ["TRIGLY", "Triglycerides", "Fasting triglyceride measurement", "Primary"],
    ["HDL", "HDL cholesterol", "Direct HDL cholesterol measurement", "Primary"],
    ["DR1TOT", "Dietary intake", "First-day total nutrient intake", "Secondary"],
    ["PAQ", "Physical activity", "Self-reported physical-activity information", "Secondary"]
], columns=[
    "Dataset",
    "Description",
    "Information_Used",
    "Analysis_Priority"
])

raw_column_inventory = []

print("Building technical column inventory")
print("=" * 70)

for number, xpt_path in enumerate(extracted_xpt_files, start=1):

    cycle = xpt_path.parent.name
    dataset_name = xpt_path.stem.upper()

    dataframe = pd.read_sas(
        xpt_path,
        format="xport",
        encoding="latin-1"
    )

    for column in dataframe.columns:

        non_missing = int(dataframe[column].notna().sum())
        missing = int(dataframe[column].isna().sum())
        unique_values = int(dataframe[column].nunique(dropna=True))

        examples = (
            dataframe[column]
            .dropna()
            .astype(str)
            .drop_duplicates()
            .head(3)
            .tolist()
        )

        raw_column_inventory.append({
            "Survey_Cycle": cycle,
            "Dataset": dataset_name,
            "Variable": column,
            "Data_Type": str(dataframe[column].dtype),
            "Total_Rows": len(dataframe),
            "Non_Missing": non_missing,
            "Missing": missing,
            "Missing_Percent": (
                missing / len(dataframe)
                if len(dataframe) > 0
                else None
            ),
            "Unique_Non_Missing": unique_values,
            "Example_Values": ", ".join(examples)
        })

    print(
        f"{number:02d}/30 | {cycle} | "
        f"{dataset_name} | {len(dataframe.columns)} variables"
    )

raw_columns_dictionary = pd.DataFrame(raw_column_inventory)

dictionary_folder = project_folder / "data_dictionary"
dictionary_folder.mkdir(parents=True, exist_ok=True)

dictionary_file = dictionary_folder / "NHANES_Data_Dictionary.xlsx"

with pd.ExcelWriter(dictionary_file, engine="openpyxl") as writer:

    research_dictionary.to_excel(
        writer,
        sheet_name="Research_Variables",
        index=False
    )

    dataset_register.to_excel(
        writer,
        sheet_name="Dataset_Register",
        index=False
    )

    raw_columns_dictionary.to_excel(
        writer,
        sheet_name="All_Raw_Columns",
        index=False
    )

    for worksheet in writer.book.worksheets:

        worksheet.freeze_panes = "A2"
        worksheet.auto_filter.ref = worksheet.dimensions

        for cell in worksheet[1]:
            cell.font = Font(bold=True, color="FFFFFF")
            cell.fill = PatternFill(
                fill_type="solid",
                fgColor="1F4E78"
            )
            cell.alignment = Alignment(
                horizontal="center",
                vertical="center",
                wrap_text=True
            )

        worksheet.row_dimensions[1].height = 30

        for column_cells in worksheet.columns:
            column_letter = column_cells[0].column_letter

            maximum_length = max(
                len(str(cell.value)) if cell.value is not None else 0
                for cell in list(column_cells)[:250]
            )

            worksheet.column_dimensions[column_letter].width = min(
                maximum_length + 2,
                45
            )

    # Percentage formatting in the raw-column inventory
    raw_sheet = writer.book["All_Raw_Columns"]

    for cell in raw_sheet["H"][1:]:
        cell.number_format = "0.0%"

print("\nDATA DICTIONARY CREATED")
print("-" * 70)
print("Workbook:", dictionary_file)
print("Research variables:", len(research_dictionary))
print("Dataset types:", len(dataset_register))
print("Raw-column records:", len(raw_columns_dictionary))
print("Status: Successfully saved to Google Drive.")
```
